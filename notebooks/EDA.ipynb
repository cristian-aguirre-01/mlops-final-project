Observamos que '`EmployeeCount`', '`Over18`' y '`StandardHours`' tienen un solo valor único, y '`EmployeeNumber`' tiene `1470` valores únicos.

## 📉 Características Categóricas

```python
object_col = []
for column in df.columns:
    if df[column].dtype == object and len(df[column].unique()) <= 30:
        object_col.append(column)
        print(f"{column} : {df[column].unique()}")
        print(df[column].value_counts())
        print("====================================")
object_col.remove('Attrition')
```

```python
from sklearn.preprocessing import LabelEncoder

label = LabelEncoder()
df["Attrition"] = label.fit_transform(df.Attrition)
```

## 📝 **Conclusiones:**

***
- Los trabajadores con bajo **`JobLevel`**, **`MonthlyIncome`**, **`YearsAtCompany`** y **`TotalWorkingYears`** tienen mayor probabilidad de dejar sus trabajos.
- **`BusinessTravel`**: Los trabajadores que viajan con mayor frecuencia tienen más probabilidades de renunciar que otros empleados.

- **`Department`**: Los trabajadores en **`Research & Development`** tienen mayor probabilidad de permanecer en la empresa que los empleados de otros departamentos.

- **`EducationField`**: Los trabajadores del campo **`Human Resources`** y **`Technical Degree`** tienen mayor probabilidad de renunciar que empleados de otras áreas de estudio.

- **`Gender`**: Los empleados **`Male`** tienen mayor probabilidad de renunciar.

- **`JobRole`**: Los trabajadores en **`Laboratory Technician`**, **`Sales Representative`** y **`Human Resources`** tienen mayor probabilidad de renunciar que quienes ocupan otros cargos.

- **`MaritalStatus`**: Los trabajadores con estado civil **`Single`** tienen mayor probabilidad de renunciar que los **`Married`** y **`Divorced`**.

- **`OverTime`**: Los trabajadores que realizan más horas extra tienen mayor probabilidad de renunciar que los demás.

# 📉 Matriz de Correlación

```python
plt.figure(figsize=(30, 30))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap="RdYlGn", annot_kws={"size":15})
```

```python
df.select_dtypes(include='number').drop('Attrition', axis=1).corrwith(df.Attrition).hvplot.barh()
```

## 📝 **Análisis de resultados de correlación (análisis de muestra):**

- El ingreso mensual (**MonthlyIncome**) está altamente correlacionado con el nivel de puesto (**JobLevel**).
- El nivel de puesto (**JobLevel**) está altamente correlacionado con los años totales trabajados (**TotalWorkingYears**).
- El ingreso mensual (**MonthlyIncome**) está altamente correlacionado con los años totales trabajados (**TotalWorkingYears**).
- La edad (**Age**) también está positivamente correlacionada con los años totales trabajados (**TotalWorkingYears**).
- El estado civil (**MaritalStatus**) y el nivel de opciones de acciones (**StockOptionLevel**) están negativamente correlacionados.
